```python
import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt
import umap
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, accuracy_score
from sklearn.linear_model import LinearRegression

torch.manual_seed(0)
np.random.seed(0)
```

## Task 1

### 1.1

```python
# As Numpy
train_in  = pd.read_csv('train_in.csv',  header=None).values
train_out = pd.read_csv('train_out.csv', header=None).values.flatten().astype(int)
test_in   = pd.read_csv('test_in.csv',   header=None).values
test_out  = pd.read_csv('test_out.csv',  header=None).values.flatten().astype(int)
```

```python
# As DataFrame
train_in_df  = pd.read_csv('train_in.csv',  header=None)
train_out_df = pd.read_csv('train_out.csv',  header=None)
test_in_df = pd.read_csv('test_in.csv',  header=None)
test_out_df = pd.read_csv('test_out.csv',  header=None)
```

```python
train_df = train_in_df.set_index(train_out_df.iloc[:, 0])
centers_df=train_df.groupby(by=train_df.index).mean()
```

```python
from scipy.spatial.distance import cdist

centers = centers_df.values          # (10, 256) numpy array
dist_matrix = cdist(centers, centers)  # (10, 10) pairwise Euclidean distances
```

```python
dist_matrix_df = pd.DataFrame(dist_matrix, index=centers_df.index, columns=centers_df.index)
```

```python
dist_matrix_df.style.background_gradient(cmap='viridis').format('{:.1f}')
```

```python
masked = np.where(dist_matrix == 0, np.nan, dist_matrix)
min_val = np.nanmin(masked)
i, j = np.where(masked == min_val)
print(f"Closest pair: digits {i[0]} and {j[0]}, distance = {min_val:.3f}")
```

We expect that there is going to be some misclasification since there are pairs with lower Distances than expected.

```python
pca_result = PCA(n_components=2).fit_transform(train_in)
tsne_result = TSNE(n_components=2, random_state=0).fit_transform(train_in)
umap_result = umap.UMAP(random_state=0).fit_transform(train_in)
```

```python
%matplotlib inline
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
scatter0 = axes[0].scatter(pca_result[:,0],pca_result[:,1],s=7,c=train_out, cmap='tab10')
axes[0].set_title('PCA')
scatter1 = axes[1].scatter(tsne_result[:,0],tsne_result[:,1],s=7,c=train_out, cmap='tab10')
axes[1].set_title('TSNE')
scatter2 = axes[2].scatter(umap_result[:,0],umap_result[:,1],s=7,c=train_out, cmap='tab10')
axes[2].set_title('UMAP')
fig.colorbar(scatter0, ax=axes, label='digit')

plt.show()
```

UMAP shows the most distinct clustering out of all the methods used. TSNE follows with good grouping, only few more overlaps than umap. Lastly PCA shows considerable overlap and not very clear clustering. From all three methods we do observe what was evident from our Euclidian Distance matrix, digits 7 and 9 (dist: 5.4) show overlap and so does 4 and 9 (dist: 6.0).

### 1.3

```python
dist_to_centers_train = cdist(train_in_df, centers_df)   # shape (1707, 10)
dist_to_centers_test  = cdist(test_in_df, centers_df)    # shape (1000, 10)

predictions_train = dist_to_centers_train.argmin(axis=1)   # (1707,) numpy array
predictions_test  = dist_to_centers_test.argmin(axis=1)    # (1000,)

print(f"Train set: Accuracy of Nearest mean classifier predicitons is {(predictions_train == train_out).mean():.2%}")
print(f"Test set: Accuracy of Nearest mean classifier predicitons is {(predictions_test == test_out).mean():.2%}")
```

### 1.4

```python
#We'll use scikit-learn's KNN
knn = KNeighborsClassifier(n_neighbors=5)# We chose k=5 as a standard default value for KNN; being odd, it also avoids the tie-breaking ambiguity that can occur with even values of k
knn.fit(train_in, train_out)
train_pred_knn = knn.predict(train_in)
test_pred_knn = knn.predict(test_in)
print(f"Train set: Accuracy of KNN predicitons is {(train_pred_knn == train_out).mean():.2%}")
print(f"Test set: Accuracy of KNN predicitons is {(test_pred_knn == test_out).mean():.2%}")
```

## Task 2

```python
def augment(T):
    return np.column_stack((T, np.ones(T.shape[0]))) #augmenting the bias

def init_weights(w):
    return np.random.randn(257, 10) * w #initialising random weights

def predict(T_aug, W):
    return (T_aug @ W).argmax(axis=1) #prediction
```

```python
def train_perceptron(T_aug, y, lr=1.0, w=0.01, epochs=100):
    W = init_weights(w)
    W_per_epoch=[]
    for epoch in range(epochs):
        preds = predict(T_aug, W)
        misclassified = np.where(preds != y)[0] #run only on the misclassified digits
        for idx in misclassified:
            t = T_aug[idx]
            true_label = y[idx]     
            pred_label = preds[idx]
            W[:, true_label] += lr * t
            W[:, pred_label] -= lr * t

        W_per_epoch.append(W.copy()) #Get the weight for each epoch -> plotting later
        
    return W,W_per_epoch
```

```python
W,trained_perceptron=train_perceptron(augment(train_in),train_out)#train the model
```

```python
# Get the train_set and test_set accuracies
print(f"Train set: Accuracy of perceptron predicitons is {(predict(augment(train_in),W)==train_out).mean():.1%}")
print(f"Test set: Accuracy of perceptron predicitons is {(predict(augment(test_in),W)==test_out).mean():.1%}")
```

```python
def Visual_Acc_per_epoch(T_aug,y,trained_perceptron,title=""):
    accs=[(predict(T_aug,W)==y).mean() for W in trained_perceptron]
    plt.plot(accs, marker="o", markersize=4)
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.title(f"{title} Perceptron Accuracy per Epoch")
    plt.grid(True)
    plt.show()


Visual_Acc_per_epoch(augment(train_in),train_out,trained_perceptron,title="Training Set")
Visual_Acc_per_epoch(augment(test_in),test_out,trained_perceptron,title="Test Set")
```

Accuracy per epoch is gradually increasing as expected. The train-set slowly converges to 100% accuracy while the test-set reaches around 86-88%.

```python
final_accs = []
for run in range(10):
    W, hist = train_perceptron(augment(train_in), train_out)
    final_accs.append((predict(augment(test_in), W) == test_out).mean())
print(final_accs)
```

We observe stability of our accuracies

```python
# We try some different parameters
W,trained_perceptron=train_perceptron(augment(train_in),train_out,lr=0.001,w=10)#train the model

print(f"Train set: Accuracy of perceptron predicitons is {(predict(augment(train_in),W)==train_out).mean():.1%}")
print(f"Test set: Accuracy of perceptron predicitons is {(predict(augment(test_in),W)==test_out).mean():.1%}")
Visual_Acc_per_epoch(augment(train_in),train_out,trained_perceptron,title="Training Set")
Visual_Acc_per_epoch(augment(test_in),test_out,trained_perceptron,title="Test Set")
```

```python
final_accs = []
for run in range(10):
    W, hist = train_perceptron(augment(train_in),train_out,lr=0.001,w=10)
    final_accs.append((predict(augment(test_in), W) == test_out).mean())
print(final_accs)
plt.plot(final_accs,marker="o")
plt.ylim(0.5,0.9)
```

We observe that with lower learning rates and increased random weight spread, it takes considerably longer for the line to converge.
